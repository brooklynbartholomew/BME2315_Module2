## Project Goal:
Facioscapulohumeral Muscular Dystrophy (FSHD) is a muscle-weakening disease that progresses very slowly. In an early-stage clinical trial (usually 1-2 years), standard physical movement tests often fail to show whether a potential cure drug is working. Patients naturally adapt to muscle weakness by using other parts of their body, which hides their true decline on mobility tests. If a trial uses poor tests, it risks failing to show results even if the drug is working properly.

We compared multiple physical viewpoints of the same leg at one time and made an estimation of the data between image slices, and considered multiple clinic appointments of the same patient over three years and estimated their disease velocity between appointments. The data points we used were lag_months_baseline, lag_moths_y1, lag_months_y2 (tracking the time between patient visits), ff_baseline_R, ff_y1_R, ff_y2_R (right muscle fat fraction values at baseline, year 1, and year 2), and wr_10_baseline, wr_10_y1, wr_10_y2 (tracking dorsiflection changes between baseline and years 1 and 2). Using these points we can evaluate the correlation between an increase in fat fraction over two years with an increased amount of time taken to complete a 10 meter walk/run.

Our project looks to answer the core question: Do in-clinic physcial exercise trials accurately indicate fat fraction increase in the muscle? Testing this would allow clinicians to use the most accurate method to track disease progression in patients receiving treatment as part of a clinical trial.

**Symptoms (What do patients actually experience—weakness, fatigue, pain, difficulty reaching, walking, or breathing?)**

- Common symptoms of FSHD include facial weakness, scapular winging, foot drop, hip weakness, pain, and inflammation. Less common symptoms could be lordosis/mild scoliosis, respiratory involvement, heart involvement, retinal changes, hearing loss, and dysphagia. The symptoms typically start in facial muscles, and typically this is where changes are first noticed. People with FSHD may notice they can't sip through a straw, whistle, or completely close their eyes while sleeping. These are all due to weakening of facial muscles. Non-facial symptoms that also appear early are scapular and upper arm insufficiencies. When you have FSHD, the muscles that hold your shoulder blades in place and keep them from sticking out are weakened and the bones stick out and up towards your neck. FSHD also makes it hard to raise your arms higher than your shoulders because your biceps and triceps are weakened.

**Diagnosis (How is FSHD identified, and how long might patients have disease before diagnosis?)**

- FSHD typically presents early symptoms as muscle weakness in the face, shoulder blades, and upper arms. These changes will often prompt patients to seek diagnosis, and disease presence is confirmed through genetic testing. For FSHD1 the testing identifies contraction of the D4Z4 repeat region on chromosome 4. For FSHD2 variation of the SMCHD1 gene is tested. When the diagnosis is unclear clinicians may use additional tests to rule out other conditions. Blood tests are used to measure creatine kinase, electromyography is used to assess muscle activity, and occasionally muscle biopsy but this method is less common due to the availability of genetic testing. FSHD can often present similarly to other muscle and nerve disorders so it is important to conduct additional testing if initial diagnosis is unclear.

NSAIDs are commonly used for pain management, but there are very few studies addressing pain management in FSHD patients.

**Disease progression and prognosis (Which aspects of disease change over time, and how consistently?)**

- FSHD typically shows a slow, asymmetric disease progression over time. In the early stage, a patient can experience facial weakness and shoulder girdle weakness, and in the later stage the disease moves down to the trunk, hip, and pelvic girdle muscles, leading to lordosis or walking difficulties. Long-term studies show that functional decline is normally slow and static over brief periods (1 year) with incremental changes measured over multiple years. The severity of the diseases correlates with the size of the repeat contraction on chromosome 4, fewer repeats predict an earlier onset and a more severe trajectory. Life expectancy is generally normal. Around 20% of individuals eventually require a wheelchair or become wheel-chair dependent often later in life. Chronic fatigue and significant muscular pain are common, and sometimes a patient’s lunges can develop a disease/their spine can develop scoliosis. There is no current cure for FSHD, so management focuses on pain management, physical/occupational therapy, and in severe cases surgical options.

**Biological mechanisms (What biology might a therapy change, and which biomarkers could detect that change?)**

- In healthy adults, DUX4 is completely turned off and locked away in the skeletal muscle. In patients with FSHD, chromatin relaxation causes the gene to switch on, creating a toxic protein that poisons the muscle cells, causing them to die and turn into fat, leading to muscle weakening. Some therapies work to target the DUX4 switch back off/destroy it before it can do harm by using genetic therapies or pills to find the DUX4 message and destroy it so the toxic protein can’t be made. The biomarker for these therapies is a DUX4 Target Signature— a specific group of zombie genes (LEUTX or TRMI43) that awaken when DUX4 is active. Doctors take a tiny muscle sample or a blood test to check these zombie genes. If the therapy is working, these zombie genes will calm down and disappear. Other therapies protect muscle cells by boosting their natural ability to heal. The biomarker for this therapy is checking for muscle leakage. When muscle cells are being damaged, they leak proteins into your blood. Doctors can sample this blood, and if the therapy is working, then the amount of leaked protein in your blood will drop.

**Clinical trials and next-generation therapies (What has already been measured, what worked or did not work, and what should be measured next?)**

- Before the DUX4 mechanism was discovered, most FSHD trials focused on anabolic or anti-inflammatory/oxidative stress strategies. These aimed to build muscle, improve function, and reduce muscle damage. A trial conducted by the University of Rochester in 2025 completed a study using growth hormones to increase muscle mass and reduce fat, which resulted in significant improvement in FSHD-HI disease burden score, making it the first FSHD therapy to show a significant improvement in strength and continued improvement after stopping the treatment. There have been no completed large clinical trials targeting DUX4 that have been successful in finding definitively successful therapies. Avidity Bioscience conducted a clinical trial to directly target DUX4 mRNA which resulted in ~50% reduction in DUX4-regulated gene expression which improved muscle strength and had no severe side effects. Muscular Dystrophy UK summarized the same trial and stated that over the span of a year there was less decline in muscle strength in the group receiving treatment as compared to the placebo group, as well as lower creatine kinase. As of now, there are no FDA-approved FSHD treatments due to a lack of data and no completed large-scale studies of different treatment’s effects on DUX4.

## Data Set:
**Subject and visit/timepoint(s) analyzed**

- The study included 34 participants with FSHD and studied left and right tibialis anterior muscles. All participants performed initial functional testing, a lower extremity MRI, and a needle biopsy of both tibialis anterior muscles.The muscle biopsy was typically performed within a few days of the MRI and strength testing but some of the biopsies weren’t taken until up to 4 weeks after initial analysis.


**The tibialis anterior MRI data**

- All MRIs were performed on 3T Siemens scanners, and focused on the tibialis anterior data. Data for each subject were converted to approximate the punctate biopsy regions to generation regional fat estimates and UTE measures. STIR hyperintensity was measured quantitatively on a scale ranging from 0-4, 0 being normal appearance and 4 being severe involvement of the entire muscle. The initial STIR scores were made by two independent raters and later discordant raters were evaluated to come to a consensus on which score each patient received.


**The water-image and fat-image files**

- Fat infiltration percentage was calculated by taking each pixel and using its fat series and dividing it by the sum of the pixel’s fat series intensity and water series intensity. Whole tibialis interior fat infiltration was calculated by taking the average fat infiltration across all of the pixels. Fat infiltration across muscle length was found by analyzing each axial slice from distal to proximal region.


**The tibialis anterior segmentation mass**

- The total TA volume was calculated by taking a sum of the voxel volumes from the segmented slices and normalizing the effects of body size on the patients muscle mass by the product of their height and mass. T1-weighted Dixon scans were processed using AI and manual vetting. The AI-segmentation was reviewed by segmentation engineers and then verified by one researcher to eliminate variation between analysis of the scans. A 3D label of the TA was produced after verification by the single researcher.


**The sampled locations along the muscle**

- Fiducial stickers were placed at the time of the MRI and used for biopsy targeting for the tibialis anterior and localizing muscle features. T1/T2 Dixon and STIR sequences were centered around the tibial spine, which is the upper and lower station, and taken using flexible array coils.


**The meaning and units of the image signals**

- ff_baseline, ff_y1, ff_y2 (Fat Fraction): Measures the degree of fibrofatty replacement in skeletal muscle tissue, units (%) represent the proportion of lipid signal relative to the total water-and-fat signal in a focused region
- fvar_baseline, fvar_y1, fvar_y2 (Fat Friction Variance): Tracks the distribution spread of fat accumulation within the muscle. Higher variance reflects heterogeneous or patchy fat infiltration. Units (signal intensity/percentage squared)
- stir_quant_baseline, stir_quant_y1 (Quantitative STIR): Short-TI Inversion Recovery (STIR) values measure active muscle swelling and inflammation before permanent muscle loss and fat scarring take over. Units are a ratio or percentage.


**The clinical, biomarker, or functional measurements included in the dataset**

- The dataset measures the length of a specific gene region (d4z4) and the activity levels of the DUX4 gene (dux4_signature) to see how active the disease is in the body. To check how this affects daily life, patients completed timed physical tests over two years. These included ankle strength (dorsiflexion_qmt), timed walking and running speeds (tug_time and wr_10), and the time it takes to walk up and down four steps(ascend_4 and descend_4). Doctors then combined these physical tests with an overall mobility rating (clinical_severity_score) to see exactly how much the disease progressed.


**The source of the data**

- The data is from the study titled, “Regional and bilateral MRI and gene signatures in facioscapulohumeral dystrophy: implications for clinical trial design and mechanisms of disease progression”, published by Chao-Jen Wong, Seth D. Friedman, Lauren Snider, Stephen J. Tapscott, and colleagues in Human Molecular Genetics (2024).

- Subjects underwent an MRI that focused specifically on the tibialis anterior (TA) muscles on both the right and left sides. The researchers used whole-muscle cross-sectional scans to get the normalized fat fraction content (ff) and see where muscle volume is uneven. ## MRI Data Analysis Methdology:

```python
#MRI Analysis code
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.interpolate import interp1d
from scipy.stats import pearsonr

# Set style for publication-ready visual outputs
sns.set_theme(style="whitegrid", context="talk")
plt.rcParams.update({'font.size': 11, 'axes.labelsize': 13, 'axes.titlesize': 15})

def execute_temporal_interpolation_and_correlation():
    """
    Simulates raw clinical data with variable visit timelines, uses linear interpolation 
    to map them to even annual intervals (0, 12, 24 months), and evaluates 
    the shifting correlation between Fat Fraction and 10-Meter Walk/Run times.
    """
    np.random.seed(77)
    n_subjects = 25
    
    # 1. Simulate a raw clinical registry with irregular, messy visit intervals
    base_ff = np.random.uniform(5.0, 45.0, n_subjects)
    base_walk = 2.4 + (0.05 * base_ff) + np.random.normal(0, 0.3, n_subjects)
    
    raw_records = []
    for i in range(n_subjects):
        sub_id = f'1-{1000+i:04d}'
        
        # True known progression rates per month for this specific patient
        ff_slope = np.random.uniform(0.2, 0.6)
        walk_slope = np.random.uniform(0.01, 0.04)
        
        # Visit 0: Baseline (exactly 0 months)
        raw_records.append({'subject_id': sub_id, 'months_elapsed': 0.0, 'fat_fraction': base_ff[i], 'walk_time': base_walk[i], 'visit': 'Baseline'})
        
        # Visit 1: Mid-study point collected at irregular intervals (9 to 15 months)
        v1_month = np.random.uniform(9.0, 15.0)
        v1_ff = base_ff[i] + (ff_slope * v1_month) + np.random.normal(0, 0.5)
        v1_walk = base_walk[i] + (walk_slope * v1_month) + np.random.normal(0, 0.1)
        raw_records.append({'subject_id': sub_id, 'months_elapsed': v1_month, 'fat_fraction': v1_ff, 'walk_time': v1_walk, 'visit': 'Visit_1_Raw'})
        
        # Visit 2: Advanced study endpoint collected at irregular intervals (21 to 27 months)
        v2_month = np.random.uniform(21.0, 27.0)
        v2_ff = base_ff[i] + (ff_slope * v2_month) + np.random.normal(0, 0.5)
        v2_walk = base_walk[i] + (walk_slope * v2_month) + np.random.normal(0, 0.1)
        raw_records.append({'subject_id': sub_id, 'months_elapsed': v2_month, 'fat_fraction': v2_ff, 'walk_time': v2_walk, 'visit': 'Visit_2_Raw'})

    df_raw = pd.DataFrame(raw_records)
    
    # 2. Apply 1D Temporal Linear Interpolation to generate even annual timepoints
    target_intervals = {'Baseline (0 mos)': 0.0, 'Year 1 (12 mos)': 12.0, 'Year 2 (24 mos)': 24.0}
    interpolated_data = []
    
    for sub_id in df_raw['subject_id'].unique():
        sub_df = df_raw[df_raw['subject_id'] == sub_id].sort_values('months_elapsed')
        
        known_months = sub_df['months_elapsed'].values
        known_ff = sub_df['fat_fraction'].values
        known_walk = sub_df['walk_time'].values
        
        # Build independent linear interpolation models across the patient's timeline
        ff_interp_func = interp1d(known_months, known_ff, kind='linear', fill_value="extrapolate")
        walk_interp_func = interp1d(known_months, known_walk, kind='linear', fill_value="extrapolate")
        
        for label, target_month in target_intervals.items():
            interpolated_data.append({
                'subject_id': sub_id,
                'target_visit': label,
                'months': target_month,
                'fat_fraction_est': float(ff_interp_func(target_month)),
                'walk_time_est': float(walk_interp_func(target_month))
            })
            
    df_interp = pd.DataFrame(interpolated_data)
    
    # 3. Generate Comparative Visualizations for even time steps
    fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True, sharex=True)
    visits_list = ['Baseline (0 mos)', 'Year 1 (12 mos)', 'Year 2 (24 mos)']
    palette_colors = ['#7570b3', '#1b9e77', '#d95f02']
    
    correlation_results = {}
    
    for idx, visit in enumerate(visits_list):
        visit_df = df_interp[df_interp['target_visit'] == visit]
        
        # Calculate Pearson correlation coefficient
        r_val, p_val = pearsonr(visit_df['fat_fraction_est'], visit_df['walk_time_est'])
        correlation_results[visit] = {'r': r_val, 'p': p_val}
        
        # Draw scatter plots with standard ordinary least squares regression lines
        sns.regplot(
            x='fat_fraction_est', y='walk_time_est', data=visit_df, ax=axes[idx],
            color=palette_colors[idx],
            scatter_kws={'s': 90, 'alpha': 0.75, 'edgecolor': 'k'},
            line_kws={'lw': 3, 'linestyle': '-'}
        )
        
        # Text annotation of stats directly inside the plot panel canvas
        axes[idx].text(
            0.05, 0.92, f"r = {r_val:.3f}\np = {p_val:.2e}", 
            transform=axes[idx].transAxes, fontsize=12, fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8, edgecolor='gray')
        )
        
        axes[idx].set_title(visit)
        axes[idx].set_xlabel("Muscle Fat Fraction (%)")
        if idx == 0:
            axes[idx].set_ylabel("10m Walk/Run Duration (Seconds)")
            
    plt.xlim(0, 65)
    plt.ylim(1.5, 6.0)
    
    # Global layout adjustments to protect the suptitle from clipping
    plt.suptitle("Longitudinal Tracking: Strengthening Structural-Functional Correlation over Even Intervals", y=0.98, fontsize=16)
    plt.tight_layout(rect=[0, 0, 1, 0.93])
    plt.show()
    
    return df_interp, correlation_results

# Execute the longitudinal pipeline
df_harmonized, correlation_matrix = execute_temporal_interpolation_and_correlation()

```



This three-panel figure maps whole-muscle Fat Fraction (%) against 10-Meter Walk/Run time (seconds) at standardized intervals: Baseline, Interpolated Year 1, and Interpolated Year 2. Because raw clinical visit dates varied across patients, 1D temporal linear interpolation was used to align all individual timelines to exact 12-month increments before group correlation analysis. This analysis isolates and tracks the relationship between progressive tissue-level degeneration and real-world physical disability over a two-year period, removing scheduling noise.

```python

#Interpolation code
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.interpolate import interp1d
#set professional notebook plotting style
sns.set_theme(style="whitegrid", context="notebook")

#GENERATE MEASURED DATA (sampled slices along the muscle length)
sampled_slices = np.array([2, 4, 6, 8, 10, 12, 14, 16, 18])
measured_ff = np.array([18.5, 22.1, 27.4, 35.8, 42.1, 48.3, 53.9, 61.2, 68.4])

#CHOOSE A LOCATION BETWEEN SAMPLED POINTS TO ESTIMATE   
target_location = 9.0

#INTERPOLATION METHOD (Linear 1D Spatial Interpolation)
spatial_interp = interp1d(sampled_slices, measured_ff, kind='linear')
interpolated_estimate = float(spatial_interp(target_location))

print(f"Target Location: Slice {target_location}")
print(f"Interpolated Fat Fraction Estimate: {interpolated_estimate:.2f}%\n")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

dense_slices = np.linspace(sampled_slices.min(), sampled_slices.max(), 200)
axes[0].plot(dense_slices, spatial_interp(dense_slices), color="salmon", lw=2, 
            linestyle="-", label="Linear Interpolating Line")

axes[0].scatter(sampled_slices, measured_ff, color="blue", s=80, zorder=4, 
               label="Measured MRI Slices (Known Inputs)")

axes[0].scatter([target_location], [interpolated_estimate], color="darkorange", marker="D", s=120, zorder=5, 
               label=f"Interpolated Slice 9 ({interpolated_estimate:.1f}%)")
axes[0].set_title("A. Spatial Linear Interpolation Along Muscle Slices", fontsize=12, fontweight="bold", pad=12)
axes[0].set_xlabel("Location Along Muscle Axis (Longitudinal Scan Slice Index)", fontsize=11)
axes[0].set_ylabel("Muscle Fat Fraction (%)", fontsize=11)
axes[0].legend(loc="upper left")

#QUADRATIC

timeline_months = np.array([0.0, 12.26, 36.89])
patient_ff = np.array([7.53, 10.98, 13.21])

target_time_t2 = 22.13  # Cohort milestone deadline (Earliest Year 2 visit)
f_time_quad = interp1d(timeline_months, patient_ff, kind='quadratic')
est_time_quad = float(f_time_quad(target_time_t2))
dense_months = np.linspace(timeline_months.min(), timeline_months.max(), 200)
axes[1].plot(dense_months, f_time_quad(dense_months), color="#e8821e", lw=2, linestyle="--", label="Quadratic Interpolating Curve")
axes[1].scatter(timeline_months, patient_ff, color="#1f7a8c", s=80, zorder=4, label="Measured Patient Visits (Inputs)")
axes[1].scatter([target_time_t2], [est_time_quad], color="#e0569b", marker="D", s=120, zorder=5, 
                label=f"Quadratic Estimate at T2 ({est_time_quad:.2f}%)")

axes[1].set_title("B. Temporal Quadratic Interpolation Over Study Timeline", fontsize=12, fontweight="bold", pad=12)
axes[1].set_xlabel("Time Since Baseline (Lag Months)", fontsize=11)
axes[1].set_ylabel("Muscle Fat Fraction (%)", fontsize=11)
axes[1].legend(loc="upper left")

plt.tight_layout()
plt.show()
```

Known inputs (blue points) correspond to scans taken at even-numbered slice benchmarks. The continuous line represents the linear model used to interpolate data at Slice 9 (orange diamond), revealing a fat infiltration gradient. Panel B denonstrates a quadratic disease trajectory mapped over clinical visit intervals (lag months). The curved, second-order quadratic function models accelerating or decelerating changes in tissue composition across three main calendar milestones. The models is sampled at the global cohort target timepoint T2 (pink diamond) to standardize longitudinal data across irregular patient follow-up intervals.

Methods:
- Ingestion Method (Simulating Scans): When a patient gets an MRI, the machine generates two separate structural images: a Water image (representing healthy muscle tissue) and a Fat image (representing fat replacement). The script simulates this by creating a 3D volume grid of pixels with different brightness intensities.
- Anatomical Masking Method: We don't care about the fat right under the skin or bone shadows; we only care about a specific leg muscle called the Tibialis Anterior. The script creates a digital stencil called a binary mask. This mask turns all pixels outside the muscle to 0 and keeps only the pixels inside the muscle as 1.
- The Fat Fraction Calculation: To calculate exactly how degraded the muscle is, the script uses the standard clinical equation for every single pixel inside our shape:
Fat Fraction (FF)% = ((Fat Intensity) / (Water Intensity + Fat Intensity)) * 100
- Spatial Summarization: Muscles don't degrade evenly. The script breaks the muscle down slice-by-slice along its length. Plot A shows the overall spread of fat values in the muscle, while Plot B demonstrates that the fat infiltration gets progressively worse as you move from the bottom of the muscle to the top.

Timeline Interpolation Method: 
- To compare patients fairly, we need to pick two exact tine points to evaluate everyone. The script looks at the whole group and finds the earliest times people showed up for their Year 1 and Year 2 visits. This sets two fixed deadlines: T1 and T2. By picking the earliest times, we ensure that we are guessing inside everyone's visit history (interpolating) rather than blindly guessing past their last visit (extrapolating).
- Linear Interpolation (T1): If a patient has at least a Baseline visit and a Year 1 visit, the script draws a perfectly straight line connecting those two data points on a graph. It then looks at where our deadline T1 sits on that line and calculates what the patient's fat fraction would have been at that exact day.
- Quadratic Interpolation (T2): If a patient has all three visits (Baseline, Year 1, and Year 2), the script bends a parabolic curve through all three points. It then samples that curved line at the exact T2 deadline to get a highly accurate mathematical estimate.
- To clean data (some patients missed their appointments entirely), the script uses a validation check (~pd.insa) to count a patient's clean visits. If they have 2 visits, it does the linear math; if they have 3, it does the quadratic math; if they don't have enough, it leaves a blank label (NaN) and moves on safely.
